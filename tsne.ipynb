```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_digits
```

```python
def tsne_dimension_reduction(data,labels=None, n_components=2,perplexity=30,learning_rate=200,random_state=42):
    scaler = StandardScaler()
    data_scaled =  scaler.fit_transform(data)

    tsne = TSNE(n_components=n_components,perplexity=perplexity,learning_rate=learning_rate,random_state=random_state)
    reduced_data = tsne.fit_transform(data_scaled)

    plt.figure(figsize=(8,6))
    if n_components == 2:
        plt.scatter(reduced_data[:,0],reduced_data[:,1],c=labels,cmap='viridis',alpha=0.7)
        plt.colorbar(label='Classes')
        plt.xlabel("TSNE Component 1")
        plt.ylabel("TSNE Component 2")
        plt.title("T-SNE Visualization (2D)")
    elif n_components == 3:
        from mpl_toolkits.mplot3d import Axes3D
        fig = plt.figure(figsize=(10,8))
        ax = fig.add_subplot(111, projection='3d')
        ax.scatter(reduced_data[:,0],reduced_data[:,1],reduced_data[:,2],c=labels,cmap='viridis',alpha=0.5)
        ax.set_xlabel("TSNE Component 1")
        ax.set_ylabel("TSNE Component 2")
        ax.set_zlabel("TSNE Component 3")
        plt.title("T-SNE Visualization (3D)")
    plt.show()
```

```python
digits = load_digits()
data,labels = digits.data, digits.target
reduced_data = tsne_dimension_reduction(data,labels, n_components=3)
```
