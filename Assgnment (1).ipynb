```
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

# Train models
lr_model = LogisticRegression()
svm_model = SVC(probability=True)   # important for ROC
nb_model = GaussianNB()

lr_model.fit(X_train, y_train)
svm_model.fit(X_train, y_train)
nb_model.fit(X_train, y_train)

# Get probabilities
y_prob_lr = lr_model.predict_proba(X_test)[:, 1]
y_prob_svm = svm_model.predict_proba(X_test)[:, 1]
y_prob_nb = nb_model.predict_proba(X_test)[:, 1]

# ROC values
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_svm, tpr_svm, _ = roc_curve(y_test, y_prob_svm)
fpr_nb, tpr_nb, _ = roc_curve(y_test, y_prob_nb)

# AUC
auc_lr = auc(fpr_lr, tpr_lr)
auc_svm = auc(fpr_svm, tpr_svm)
auc_nb = auc(fpr_nb, tpr_nb)

# Plot
plt.figure(figsize=(7,5))

plt.plot(fpr_lr, tpr_lr, label=f"Logistic (AUC = {auc_lr:.2f})")
plt.plot(fpr_svm, tpr_svm, label=f"SVM (AUC = {auc_svm:.2f})")
plt.plot(fpr_nb, tpr_nb, label=f"Naive Bayes (AUC = {auc_nb:.2f})")

plt.plot([0,1], [0,1], linestyle='--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()

plt.show()
```

#1. Which model performed best and why?

In my project, Logistic Regression performed the best overall.
It achieved high accuracy along with a good balance between precision, recall, and F1 score.
Unlike other models, its performance was stable and consistent across the dataset.
Even though Naive Bayes showed slightly higher accuracy, it is based on strong assumptions.
Logistic Regression is more reliable and works well in real-world scenarios.
Therefore, I considered Logistic Regression as the best model.

#2. Why did Naive Bayes perform poorly?

Naive Bayes is based on the assumption that all features are independent of each other.
In real-world datasets, features are usually correlated, which violates this assumption.
Because of this, the model may not capture the true relationships in the data.
It can sometimes give high accuracy but may not generalize well.
This makes it less reliable compared to other models like Logistic Regression.
Hence, its performance may not always be trustworthy.

#3. Compare SVM performance before and after scaling

Before applying scaling, SVM showed lower accuracy and unstable results.
This is because SVM relies on distance between data points.
When features are not scaled, some features dominate others, affecting performance.
After scaling, all features are normalized and contribute equally.
This leads to better decision boundaries and improved accuracy.
Therefore, scaling significantly improves the performance of SVM.

#4. Which model trained fastest and why?

Naive Bayes was the fastest model during training.
This is because it uses simple probability calculations instead of complex optimization.
It does not require iterative learning like Logistic Regression or SVM.
Logistic Regression takes moderate time due to parameter optimization.
SVM takes the longest time, especially with non-linear kernels like RBF.
Hence, Naive Bayes is the fastest among all three models.

#5. Which model results were easiest to explain?

Logistic Regression was the easiest model to explain.
It is based on a simple linear relationship between input features and output.
The model provides clear and interpretable results.
It is easier to understand how each feature affects the prediction.
Compared to SVM and Naive Bayes, it is more transparent.
Therefore, Logistic Regression is more suitable for explanation and real-world use.
